```python
# Find the headers
headers = [th.text.strip() for th in table.find("thead").find_all("th")]

headers
```

```python
row_data = []

if table:
    for tr in table.find("tbody").find_all("tr"):
        cells = [td.text.strip() for td in tr.find_all("td")]
        row_data.append(cells)

row_data
```

```python
#Combine Headers and Data into a DataFrame
web_df = pd.DataFrame(row_data, columns=headers)

web_df.head()
```

```python
# STEP 1: Missing Data

# Check if there are any missing values
print(web_df.isnull().values.any())
```

```python
# STEP 2: Check Data Types

#Check the data type of entire DataFrame:
print(web_df.dtypes)
```

```python
# STEP 3: Convert Data Types
# Since currency is being stored as an object, convert it to float for ability to analyze

cols_to_convert = web_df.columns[1:]  # The first column is "Country Name" and should remain unchanged

# First remove the commas and dollar sign from the column
web_df[cols_to_convert] = web_df[cols_to_convert].apply(lambda x: x.str.replace(",", "").str.replace("$", ""))

# Now convert the results to a float
web_df[cols_to_convert] = web_df[cols_to_convert].astype('float')

#Check the data type of entire DataFrame:
print(web_df.dtypes)
```

```python
# STEP 4: Convert Data Types
# Now convert the Country Name column to a string
web_df['Country Name'] = web_df['Country Name'].astype("string")

#Check the data type of entire DataFrame:
print(web_df.dtypes)
```

```python
#STEP 5: MELT
# To match the format of my other data sources, pivot from "wide" to "long" format

web_df_melted = web_df.melt(id_vars = ["Country Name"],
                            var_name = "Year",
                            value_name = "GDP")

# View results for US to see if it worked:
web_df_melted.loc[web_df_melted["Country Name"] == "United States"]
```

```python
#Check the data type of entire DataFrame:
print(web_df_melted.dtypes)
```

```python
#STEP 6: Update data type again
# Now update datatype for Year from an object to int.
web_df_melted['Year'] = web_df_melted['Year'].astype("int")

print(web_df_melted.dtypes)
```

```python
# STEP 7: Find any duplicates
# Check for any duplicated values across all columns.
web_df_melted[web_df_melted.duplicated()]
```

```python
# STEP 8: Handle $0 GDP Values
# There may be various reasons the data has $0 as a GDP value (Missing data, data errors, extremely small economies, extreme crises)
# However regardless of the reason, in reality, a country's GDP cannot exactly be $0.
# So in this step, we will convert $0 to NaN to prevent any misleading analysis.

# Find rows with GDP value of $0
zero = web_df_melted[web_df_melted["GDP"] ==0]
print(zero)
```

```python
#Replace the 0 values with NaN
import numpy as np
web_df_melted["GDP"] = web_df_melted["GDP"].replace(0, np.nan)

# Check for 0s again
zero = web_df_melted[web_df_melted["GDP"] ==0]
print(zero)
```

# Ethical Implications  

In the data transformation and cleaning process applied to the MacroTrends GDP website data, several steps were taken to enhance the dataset's usability and accuracy. First, the data was extracted from an online table, and currency formatting was removed to convert the GDP values into a numeric format. This was done so that, in future steps, numerical analysis could be conducted on these values. Additionally, the dataset was reformatted from a wide format to a long format to align with a future step in which we will join this data with that from Milestone 2, which follows the same format. Finally, another significant step taken was to replace any GDP values represented as `$0` with NaN in an effort to eliminate misleading analysis. This was an assumption taken during this phase of the project that should be thoroughly documented (as demonstrated in the code above). Treating `$0` GDP values as missing data, which, if incorrect, could impact any downstream analyses.
Unfortunately, one ethical concern arose regarding data ownership and programmatic access upon reviewing the site's terms of service. MacroTrends discourages automated extraction of data, which raises legal concerns. To mitigate this ethical risk, alternative open-data sources should be considered in the future, especially if the data is used beyond a school project, to ensure compliance with data access policies.
As for the credibility of this data source, the MacroTrends website sources its data directly from The World Bank, which is a highly credible and authoritative source for GDP data. The World Bank ensures high-quality data by using methodologies that are internationally accepted and encourages countries to improve the quality of their official statistics. However, it is important to note that there may be a lag in the availability of recent data due to the time required for data collection and validation from various countries. I believe this is evident in the `$0` GDP values, discussed earlier. For the most up-to-date data, official national websites may be a valuable resource to explore.
